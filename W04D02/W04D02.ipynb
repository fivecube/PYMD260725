```python
import time

def greet_decorator(func):
    def wrapper():
        start_at = time.time()
        print("hello")
        func()
        print('Goodbye!')
        end_at = time.time()
        with open('logging.txt', 'w') as fp:
            fp.write(f'{func.__name__} took time {end_at-start_at}')
    return wrapper
```

```python
@greet_decorator
def say_name_kavita():
    print('I am Kavita!')
```

```python
say_name_kavita()
```

```python
Use cases of Decorators:-
1. Logging
2. Authentication and checking Permissions
3. Cache Results
4. Rate Limiting
```

```python
# Context managers

class MyContext:
    def __enter__(self):
        print('Entering..')
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        print('Exiting...')
```

```python
with MyContext():
    print('I am inside the context!')
```

```python
# Custom Context Manager

class CustomOpen:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.file.close()
        return False
```

```python
with CustomOpen("logging.txt", "r") as fp:
    content = fp.read()
    print(content)
```

```python
with CustomOpen("logging.txt", "r") as fp:
    content = fp.read()
    print(content)
```

```python
import re
```

```python
# This is a regular expression:- r"K[a-z]*a$"

list_of_text = ["We have a student named Kavita", "Karma Pays", "Kamla is not a student",
                "Kat is a good actress"]

for text in list_of_text:
    finding = re.search(r"K[a-z]*a\b", text)
    print(finding)
```

```python
# This is a regular expression:- r"K[a-z]*a$"

list_of_text = ["Kavita", "Karma", "Kamla", "Kat", "Amla"]

for text in list_of_text:
    print(text)
    finding = re.search(r"^K[a-z]*a$", text)
    print(finding)
```

```python
list_of_text = ["KT", "KaT", "K5T", "KxT", "MT", "KM"]

for text in list_of_text:
    finding = re.search(r"K[a-z]*T", text)
    print(finding)
```

```python
list_of_text = ["KT", "KaT", "K5T", "KxT", "MT", "KM"]

for text in list_of_text:
    finding = re.search(r"K[a-z]+T", text)
    print(finding)
```

```python
list_of_text = ["color", "colour", "colouur", "coloxr"]

for text in list_of_text:
    finding = re.search(r"colou?r", text)
    print(finding)
```
